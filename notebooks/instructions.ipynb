### 4. Enjoy the Simulation + Interact with the MultiSpectator Dashboard to set new Observation Targets:
The Simulation consists of 5 R-SUTs (Robotic Systems Under Test) and 8 Monitoring Robots (R0-R7). Only the Monitoring Robots are controlled by the MultiSpectator via the Dashboard. R-SUT5 consists of 3 robots and shows that the system architecture is capable to monitor multi-robot systems too. Initially the Monitoring Robots perform random exploration in the field to search and update the R-SUTs locations. All detected R-SUTs appear in the "Discovered Robots" list in the dashboard. Each of them can get a specified number of observers applied via the action column. All existing monitoring teams are displayed in the "Monitoring" table of the dashboard. This table additionally offers the possibility to stop and cancel an ongoing monitoring task. The dashboard updates every second, that is why applying an observer count has to be done quickly.  
For use case 5, fixed-position inspection, the dashboard has 3 input fields on the right, where the user can specify a target position and the number of observers for that inspection task. After pressing "Apply", the Multispectator automatically assign robots which drive to the target position and inspect it like a R-SUT (in a hexagon shape). 
If all robots are busy in monitoring teams, no new task can be applied.
